## 2. loc vs iloc
1. loc으로 인덱스 10~15행 sex, age열을 선택해 출력

3. 두 결과의 행 수를 각각 출력하고, 왜 다른지 주석 서술

```python
print(titanic.loc[10:15, "sex":"age"])
print("-----------------")
print(titanic.iloc[10:16, 2:4])
# loc은 이름표로 행 하나, 행 범위, 열 목록, 열 이름 범위를 차례로 선택함
# iloc은 위치 번호로 위치 범위, 마지막 행, 원하는 위치만 골라 선택함
# loc은 이름을 기반으로 데이터를 선택하기 때문에 위치 번호의 끝 항목을 포함하지만
# iloc은 위치 번호(정수 기반)를 기준으로 데이터를 선택하기 떄문에 범위를 선택했을 떄 끝 번호를 제외하고 표시함
```

## 5. 복합 조건
각각 조건을 만족하는 행 수를 출력하세요.
1. 3등석 그리고 남성 그리고 생존한 승객(&)
2. 1등석 또는 요금 100 이상인 승객 (|)
3. 혼자 탑승하지 않은 승객 (~, alone 열 사용)

```python
x = (titanic["pclass"] == 3) & (titanic["sex"] == "male") & (titanic["survived"] ==1 )
y = (titanic["pclass"] == 1) | (titanic["fare"] >= 100)
z = ~ (titanic["alone"])
z2 = (titanic["alone"] == False)
print("3등석, 남성, 생존 승객을 포함하는 행 수:", (len(titanic[x])))  # x: 1. 3등석 그리고 남성 그리고 생존한 승객(&)
print("-----------------")
print("1등석 또는 요금 100 이상인 승객을 포함하는 행 수: ",(len(titanic[y])))  # y: 2. 1등석 또는 요금 100 이상인 승객 (|)
print("-----------------")
print("혼자 탑승하지 않은 승객을 포함하는 행 수: ", (len(titanic[z])))  # z & z2: 3. 혼자 탑승하지 않은 승객 (~, alone 열 사용)
print("혼자 탑승하지 않은 승객을 포함하는 행 수: ", (len(titanic[z2])))
```

## 6. isin · between
1. isin 으로 Cherbourg 또는 Queenstown에서 탑승환 승객을 골라 shape를 출력하세요.
2. between으로 나이 20~30세(양 끝 포함) 승객을 골라 shape와 나이의 min·max를 출력하세요.
3. 같은 범위를 양 끝 미포함(inclusive="neither")으로 바꿔 shape를 출력하세요.

```python
mask = titanic["embark_town"].isin(["Cherbourg", "Queenstown"])
print(titanic[mask].shape)                                      # Cherboug or Queenstown에서 탑승한 승객의 shape
print("-----------------")
mask2 = titanic["age"].between(20,30, inclusive = "both")
print(titanic[mask2].shape)                                     # 2. between으로 나이 20~30세 사이(양 끝 포함) 승객의 shape
print(titanic[mask2]["age"].min(), titanic[mask2]["age"].max()) # 2. between으로 나이 20~30세 사이(양 끝 포함) 승객의 최소값 · 최대값
print("-----------------")
mask3 = titanic["age"].between(20, 30, inclusive = "neither")
print(titanic[mask3].shape)                                     # 3. between으로 나이 20~30세 사이(양 끝 미포함) 승객의 shape
```

## 7. loc[조건, 열]
12살 이하 그리고 3등석 승객의 age, sex, survived 열만 kids 변수에 담고, 행 수와 생존률(survived 평균, 소수 둘째자리)을 출력하세요

```python
mask = (titanic["age"]<=12) & (titanic["pclass"]==3)
kids = titanic.loc[mask, ["age","sex","survived"]]
print("kids 변수의 행 수: ",(len(kids)))                             # kids 변수의 행 수
survival_rate = kids["survived"].mean()
print("생존률: ",(round(survival_rate, 2)))               # 생존률(소수 둘째자리)
```

## 10. 종합 20~40세 남성, 1·2등석 vs 3등석
질문: "20~40세 남성 승객의 생존률은 객실 등급에 따라 얼마나 다를까?"
1. 남성·나이 20~40세(between)·1·2등석(isin) 조건으로 pclass, age, fare, survived 열을 선택하세요
2. 행 수, 생존률(소수 둘째 자리), 요금 상위 3명을 출력하세요.
3. 같은 조건을 query로 써서 행 수를 검증하세요
4. 등급만 3등석으로 바꿔 생존률을 구하고, 두 생존률을 비교하는 해석을 마크다운 셀에 2~3문장으로 쓰세요.

```python
mask = (titanic ["sex"] == 'male') & (titanic ["age"].between(20,40)) & (titanic["pclass"].isin([1,2]))
survivor_2040m_12 = titanic[mask][["pclass", "age", "fare", "survived"]]
print("1·2등석 남성 2040 이용객 행 수: ", len(survivor_2040m_12))
print("1·2등석 남성 2040 이용객 생존률: ", round(survivor_2040m_12["survived"].mean(),2))
print("------------------------")
print("요금 상위 3명")
print(survivor_2040m_12.sort_values(by = "fare", ascending = False).head(3))
print("-----------------------------------------------------------------")
query_survivor="sex == 'male' and age.between(20, 40) and pclass in [1, 2]"
print("query 문 행 수: ", len(titanic.query(query_survivor)))
print("-----------------------------------------------------------------")
mask3 = (titanic ["sex"] == 'male') & (titanic ["age"].between(20,40)) & (titanic["pclass"]==3)
survivor_2040m_3 = titanic[mask3][["pclass", "age", "fare", "survived"]]
print("3등석 남성 2040 이용객 행 수: ", len(survivor_2040m_3))
print("3등석 남성 2040 이용객 생존률: ", round(survivor_2040m_3["survived"].mean(),2))


print("============================================================")
print("번외 비교")
mask_f = (titanic ["sex"] == 'female') & (titanic ["age"].between(20,40)) & (titanic["pclass"].isin([1,2]))
survivor_2040fm_12 = titanic[mask_f][["pclass", "age", "fare", "survived"]]
print("1·2등석 여성 2040 이용객 행 수: ", len(survivor_2040fm_12))
print("1·2등석 여성 2040 이용객 생존률: ", round(survivor_2040fm_12["survived"].mean(),2))
print("============================================================")
mask_f3 = (titanic ["sex"] == 'female') & (titanic ["age"].between(20,40)) & (titanic["pclass"]==3)
survivor_2040fm_3 = titanic[mask_f3][["pclass", "age", "fare", "survived"]]
print("3등석 여성 2040 이용객 행 수: ", len(survivor_2040fm_3))
print("3등석 여성 2040 이용객 생존률: ", round(survivor_2040fm_3["survived"].mean(),2))
```

## 타이타닉 객실 등급에 따른 20~40 남성 생존률의 차이
1.2등석 객실을 이용한 20~40 남성의 생존률은 103명 중 24%이고 
3등석 객실을 이용한 20~40 남성의 생존률은 159명 중 15%에 불과하다. 


더 많은 인원임에도 오히려 생존률이 더 적었던 것은 높은 사회적 지위와 재력을 지닌 1,2등석 이용객들이 생존에 있어서 정보력, 구명활동 등 방면에 더 나은 영향력을 행사했다고 볼 수도 있으나 

해당 남성 집단 간의 비교가 아닌 객실별 남성과 여성의 생존률을 비교했을 떄 
1·2등석 여성 2040 이용객 생존률이 90명 중 94%
3등석 여성 2040 이용객 생존률이 48명 중 46%
라는 상대적으로 굉장히 높다는 것을 확인 할 수 있다.

이와 비교하여 객실 등급으로 인한 계층별 유리함을 논하기에는 전반적인 20~40 남성들의 생존률은 어느 비교집단이 월등하게 높다고 보기 어려우며 이들의 저조한 생존률을 보면 상대적 약자였던 여성, 노년층, 아동 등을 위해 계급여하에 상관없이 많은 양보와 희생을 했다고 볼 수 있다.
